```python
import pandas as pd

movies = pd.read_csv("../data/raw/movies.csv", encoding="latin-1")
ratings = pd.read_csv("../data/raw/ratings.csv", encoding="latin-1")
users = pd.read_csv("../data/raw/users.csv", encoding="latin-1")
```

```python
print("Unique users:", ratings["user_id"].nunique())
print("Unique movies in ratings:", ratings["movie_id"].nunique())
print("Total ratings:", len(ratings))
```

```python
total_possible = users["user_id"].nunique() * movies["movie_id"].nunique()
actual_interactions = len(ratings)

sparsity = 1 - (actual_interactions / total_possible)

print("Total possible user-movie interactions:", total_possible)
print("Actual ratings:", actual_interactions)
print("Sparsity:", sparsity)
print("Density:", 1 - sparsity)
```

```python
users["user_id"].nunique() * movies["movie_id"].nunique()
```

```python
rated_movies = ratings["movie_id"].nunique()

possible_rated_interactions = (
    ratings["user_id"].nunique() * rated_movies
)

sparsity_rated_movies = 1 - (
    len(ratings) / possible_rated_interactions
)

print("Users:", ratings["user_id"].nunique())
print("Movies with ratings:", rated_movies)
print("Possible interactions:", possible_rated_interactions)
print("Actual ratings:", len(ratings))
print("Sparsity:", sparsity_rated_movies)
print("Density:", 1 - sparsity_rated_movies)
```

```python
print("Ratings per user:")
print(ratings.groupby("user_id").size().describe())

print("\nRatings per movie:")
print(ratings.groupby("movie_id").size().describe())
```
